**### Problema 1: Cálculo da Massa Molar de uma Mistura (Listas e Laços)**
Imagine que você tem uma lista com as massas molares dos componentes presentes em uma corrente gasosa e outra lista com as suas respectivas frações molares.
O que fazer: Escreva um programa que utilize um laço for (ou uma estrutura equivalente) para calcular a massa molar aparente (média ponderada) da mistura gasosa ($\text{MM}_{mistura} = \sum (x_i \cdot MM_i)$) e exiba o resultado final.

```python
massas_molares = [16.04, 28.01, 44.01]  # Ex: Metano, Monóxido de Carbono, Dióxido de Carbono (g/mol)
fracoes_molares = [0.5, 0.3, 0.2]       # Frações correspondentes (soma = 1.0)

media_MM = sum(m * f for m, f in zip(massas_molares, fracoes_molares)) # a função zip percore duas listas.

print(f' A massa molar aparente da mistura gasosa é de {media_MM}')
```

### **Problema 2: Verificador de Limites de Pressão (Dicionários e Condicionais)**
Em uma unidade de destilação, diferentes colunas operam com pressões máximas admissíveis distintas.
O que fazer: Crie um programa que peça para o usuário digitar o nome de uma coluna e a pressão atual de operação dela. O programa deve verificar no dicionário se a coluna existe e se a pressão informada ultrapassa o limite cadastrado, exibindo um aviso de segurança ("Pressão acima do limite!") ou de operação segura ("Operação dentro dos parâmetros").

**DECOMPOSIÇÃO**:

  1. inserir coluna;
  2. inserir pressão da coluna;
  3. fazer a lógica de comparação;
  4. mostrar a resposta

**ALGORITMO**:

1. Cadastrar os limites máximos de cada coluna.
2. Pedir para o usuário digitar a coluna e a pressão atual.
3. Verificar se a coluna digitada existe no dicionário.
4. Se existir, comparar a pressão informada com o limite cadastrado.
5. Imprimir a mensagem de alerta ou de operação segura.

  

```python
limites_pressao = {
    'coluna_A': 20.3,
    'coluna_B': 50.5,
    'coluna_C': 80.2
}

coluna = input('digite o nome da coluna de destilação: ').strip()
pressao_atual = float(input('digite a pressão atual da coluna de destilação: '))

if coluna in limites_pressao:

  limites_maximo = limites_pressao[coluna]

  if pressao_atual > limites_maximo:

    print(f'A {coluna} de destilação está operando atualmente acima do limite máximo, que é de {limites_maximo}')
  else:
    print(f'A coluna {coluna} de destilação está operando atualmente a {pressao_atual} abaixo do limite, que é de {limites_maximo}')
else:
  print(f' A coluna de destilação não foi encontrada')


```

### **Problema 3: Função de Conversão de Temperatura em Lote (Funções e Listas)**:
Para monitorar reações exotérmicas, os dados de temperatura muitas vezes precisam ser convertidos de Celsius para Kelvin ou Fahrenheit. O que fazer:Crie uma função chamada converter_temperaturas(lista_celsius) que receba uma lista de temperaturas em graus Celsius.Dentro da função, utilize um laço for para converter cada valor para Kelvin ($K = \text{Celsius} + 273.15$) e armazene os novos valores em uma nova lista.A função deve retornar a nova lista convertida.No programa principal, teste a função com uma lista de temperaturas de um reator (ex: [25.0, 50.5, 80.0, 100.2]) e imprima o resultado.

**ALGORITMO:**

1. Receber uma lista de temperaturas em Celsius;
2. Criar uma nova lista vazia para armazenar as temperaturas convertidas;
3. Para cada temperatura da lista original, repetir:
   - Aplicar a fórmula de conversão (Kelvin = Celsius + 273.15);
   - Guardar o valor convertido na nova lista;
4. Retornar a nova lista pronta;
5. No programa principal, criar a lista de teste do reator,
   chamar a função e imprimir o resultado na tela.1.

```python
def converter_temperatura(lista_celsius):
  temperaturas_kelvin = []

  for T in lista_celsius:
    temp_kelvin = T + 273.15
    temperaturas_kelvin.append(temperaturas_kelvin)

  return temperaturas_kelvin

leituras_reator = [25.0, 50.5, 80.0, 100.2]

resultado_convertido = converter_temperatura(leituras_reator)
```

**🧪 Exercício Prático: Cálculo de Vazão Volumérica (Lei de Bernoulli / Hidráulica):**

Imagine que você está projetando o escoamento de um líquido por uma tubulação e precisa de uma função rápida para calcular a vazão volumétrica ($Q$), que depende da velocidade do fluido ($v$) e da área transversal da tubulação ($A$).A fórmula básica é:$$Q = A \times v$$Enunciado:Escreva um programa em Python contendo:Uma função chamada calcular_vazao que receba dois parâmetros: area (área da seção transversal em $m^2$) e velocidade (velocidade do fluido em $m/s$).Dentro da função, calcule a vazão multiplicando a área pela velocidade e armazene o resultado em uma variável.Utilize o comando return para devolver esse valor calculado para fora da função.No programa principal (fora da função):Solicite ao usuário que digite os valores da área e da velocidade (convertendo para float).Chame a função calcular_vazao passando esses valores como argumentos.Exiba o resultado final da vazão na tela de forma clara e formatada.

```python
def calcular_vazao(area, velocidade):
  vazao = (area * velocidade)
  return vazao

area = float(input('DIgite o valor da área (em m²): '))
velocidade = float(input('Digite o valor da velocidade do flúido( em m/s): '))

resultado_vazao = calcular_vazao(area, velocidade)
print(resultado_vazao)

```

🧪 Exercício: Filtrando Reatores Críticos por Temperatura

Imagine que você está monitorando vários reatores em uma planta e quer criar uma função automatizada para isolar apenas aqueles que ultrapassaram um limite crítico de temperatura.

Enunciado:
Escreva um programa em Python contendo:

Uma função chamada filtrar_reatores_quentes que receba dois parâmetros:

dados_reatores (um dicionário onde as chaves são os nomes dos reatores e os valores são as temperaturas em °C).

limite_critico (um número que define a temperatura limite de segurança).

Dentro da função, utilize uma lista de compreensão (ou um laço for tradicional) para percorrer o dicionário e coletar apenas os nomes dos reatores cuja temperatura seja maior ou igual ao limite_critico.

A função deve retornar essa lista com os nomes dos reatores críticos.

No programa principal (fora da função):

Crie um dicionário de teste com pelo menos quatro reatores e suas temperaturas (ex: {"Reator_1": 70.0, "Reator_2": 85.5, "Reator_3": 62.0, "Reator_4": 91.2}).

Defina uma variável com o valor limite (ex: 80.0).

Chame a função passando o dicionário e o limite, guarde o resultado em uma variável e exiba-o na tela com um print().

```python
def filtrar_reatores_quentes(temperaturas_reatores, limite_critico):
  return [nomes for nomes, temperaturas in temperaturas_reatores.items() if temperaturas >= limite_critico]

temperaturas_reatores = {
    "Reator_1": 70.0,
    "Reator_2": 85.5,
    "Reator_3": 62.0,
    "Reator_4": 91.2
    }

limite_critico = 80.0

reatores_quentes = filtrar_reatores_quentes(temperaturas_reatores, limite_critico)

print(reatores_quentes)

```

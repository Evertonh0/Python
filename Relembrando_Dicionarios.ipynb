### 1. Ficha Técnica de Substâncias
Crie um dicionário chamado composto que armazene as seguintes informações sobre a Água:NomeFórmula MolecularMassa Molar ($18.015$ g/mol) Ponto de Ebulição ($100$ °C)

**Desafio**: Após criar o dicionário, imprima apenas o valor da Massa Molar e adicione uma nova chave chamada estado_fisico com o valor "Líquido".

```python
# Nesse caso, a chave do dicionario é string "Água"
# variavel = {chave:valor}

composto = {
    "nome": "água",
    "Fórmula_molecular": "H2O",
    "Massa_molar (g/mol)": 18.015,
    "Ponto_Ebulição (°C)": 100
}

# Ajeitando como lista:

for i in composto.items():
  print(i)

# Printando apenas a massa molar, conforme desafio.

print(f"A massa molar da {composto['nome']} é {composto['Massa_molar (g/mol)']} g/mol")
```

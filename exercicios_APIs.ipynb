### 🧪 **Exercício Básico: Consultando Dados de Endereço via API**

Escreva um programa em Python que:

1) Importe a biblioteca requests.

2) Crie uma variável chamada url contendo o endereço da API de consulta de CEP da BrasilAPI para o CEP 01001000 (Praça da Sé, SP): [https://brasilapi.com.br/api/cep/v1/01001000](https://brasilapi.com.br/api/cep/v1/01001000)

3) Faça uma requisição GET usando requests.get().

4) Verifique se a requisição foi bem-sucedida (ou seja, se o status_code é igual a 200).

5) Se for bem-sucedida, converta a resposta para um dicionário Python usando .json() e exiba o estado (state) e a cidade (city) retornados pela API. Caso contrário, imprima uma mensagem de erro informando o código do status.

```python
# 1) Importando a biblioteca:
import requests as req

# 2) criando a variavel como string:
url = 'https://brasilapi.com.br/api/cep/v1/01001000'

# 3) solicitando requisição:
resp = req.get(url)

# 4) Vendo se a conexão deu bom ou não.
if resp.status_code == 200:
  print('conexão bem sucedida!')
else:
  print('Erro na conexão!')

# 5) Convertendo os dados para um dicionario:
if resp.status_code == 200:
  dados = resp.json()

  # 6) Exibindo a cidade e o estado:
  print(f"Cidade: {dados['city']}")
  print(f"Estado: {dados['state']}")

```
